```python
import torch
import torch.nn as nn
import torch.nn as nn
import torch.optim as optim
import numpy as np
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]
              ])
Y = np.array([[0],
              [1],
              [1],
              [0]])
```

```python
X_tensor = torch.FloatTensor(X)
y_tensor = torch.FloatTensor(Y)

class XORnet(nn.Module):
  def __init__(self):
   super(XORnet,self).__init__()
   self.hidden = nn.Linear(2,4)
   self.output = nn.Linear(4,1)
   self.sigmoid = nn.Sigmoid()


  def forward(self, x):
        x = self.sigmoid(self.hidden(x))
        x = self.sigmoid(self.output(x))
        return x


torch.manual_seed(42)
model = XORnet()

# Loss function and optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=2.0)


pytorch_loss_history = []


for i in range(10000):
  prediction = model(X_tensor);
  loss = criterion(prediction,y_tensor)
  pytorch_loss_history.append(loss.item())
  # Backward pass (automatic!)
  optimizer.zero_grad()  # Clear previous gradients
  loss.backward()        # Compute gradients (THIS IS BACKPROP!)
  optimizer.step()       # Update weights


```

```python
test = torch.FloatTensor([[0,1]])

model.eval()

with torch.no_grad():
    output = model(test)

print("Probability:", output.item())

prediction = 1 if output.item() >= 0.5 else 0

print("Prediction:", prediction)
```
